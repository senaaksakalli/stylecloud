```python
import PIL.ImageDraw
import stylecloud as sc

if not hasattr(PIL.ImageDraw.ImageDraw, 'textsize'):
    def textsize_patch(self, text, font=None, **kwargs):
        bbox = self.textbbox((0, 0), text, font=font, **kwargs)
        return bbox[2] - bbox[0], bbox[3] - bbox[1]
    PIL.ImageDraw.ImageDraw.textsize = textsize_patch
```
